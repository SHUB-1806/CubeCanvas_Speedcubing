# Question 4: Interpolation Investigation (Hard)

Different interpolation techniques produce different results.


## Part A (Theory)

Research and explain:

1. Nearest Neighbor Interpolation
2. Bilinear Interpolation
3. Bicubic Interpolation
4. Lanczos Interpolation

For each method discuss:
- Working principle
- Advantages
- Disadvantages
- Suitable applications


### Theory Answer Space

Write your answer here.

1)Nearest Neighbor Interpolation:

Nearest Neighbor interpolation is the simplest image resizing technique. It selects the value of the nearest pixel without calculating new intermediate values.

Formula : I′(x,y)=I(round(x),round(y))

Principle:Finds the nearest pixel to the new coordinate and copies that pixel value directly.No averaging or smoothing is performed.

Advantages: Fast, preserve sharp edges and is good for pixel art.

Disadvantages:Produces blocky or jagged images ,low image quality for photographs and can create visible pixelation.

Suitable Application:Pixel art,retro game graphics,fast image scaling and
binary or label images.

2)Bilinear Interpolation:

Bilinear interpolation uses four neighboring pixels to estimate a new pixel value.

Principle: Uses the nearest 2 × 2 neighboring pixels and calculates weighted averages based on distance to produce smoother transitions.

Advantages: Smoother than nearest neighbor, reduces jagged edges, faster than bicubic and Lanczos, and has moderate computational cost.

Disadvantages: Produces slight image blurring, may lose fine details, and gives lower quality compared to bicubic and Lanczos interpolation.

Suitable Application: Basic image resizing, video scaling, real-time graphics, and mobile applications.

3)Bicubic Interpolation:

Bicubic interpolation uses sixteen neighboring pixels to estimate a new pixel value.

Principle: Uses a 4 × 4 neighborhood of pixels and applies cubic interpolation calculations to produce smoother gradients and edges.

Advantages: Better image quality, smooth transitions, reduced artifacts, and sharper output than bilinear interpolation.

Disadvantages: Slower than nearest neighbor and bilinear interpolation, computationally expensive, and may create ringing artifacts.

Suitable Application: Photo editing, professional graphics software, high-quality image scaling, and printing applications.

4)Lanczos Interpolation:

Lanczos interpolation is an advanced resizing technique based on the sinc function.

Formula:
sinc(x)=sin(πx)/πx
	​

Principle: Uses multiple surrounding pixels and applies sinc-based weighting calculations to preserve sharpness and fine image details.

Advantages: Excellent image quality, sharp details, minimal aliasing, and preserves textures well.

Disadvantages: Computationally expensive, slower processing speed, and may produce ringing effects near edges.

Suitable Application: Professional image processing, high-quality photo resizing, image restoration, and publishing or printing.
